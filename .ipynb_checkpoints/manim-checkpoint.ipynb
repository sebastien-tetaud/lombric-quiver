# Manim 

The objective of the current notebook is generated animation by using essentially the library Manim. 

Manim was created by Grant Sanderson from the [3blue1brown](https://www.youtube.com/watch?v=aircAruvnKk) channel. He decided to code some visual graphs in order to animate his mathematicals content and since the open-source release, the community took it and so on is being maintained by the community with a current stable version at 0.19.

Manim is made by three main building blocks: MObjects (Mathematical Objects); Scene and Animation. Each of these buildings blocks are essentially classes. 
- MObjects is an class and holds the vector graphic information to be displayed in the scene. 
- Scene is a class and essentially serve to manage mobjects, animations and global scene configuration. 
- Animation: Any property of a mobject that can be changed can be animated. Any method that changes a mobject’s property can be used as an animation.

```python
import xarray as xr
def create_wind_xrdataset(ds:xr.Dataset, variables, date_range, spatial_range,timestep:int,
                      convert_longitude:bool=True, include_wind_speed:bool=True, include_attributes:bool=True):
    """
    Create an xarray Dataset with processed meteorological data.
    
    Parameters:
    -----------
    ds : xarray.Dataset
        The source dataset containing meteorological variables and coordinates
    variables : dict or list
        Dictionary of data variables to include or list of variable names.
        If dict, keys are variable names in the source dataset and values are new names.
        If list, the same variable names will be used in the output.
    date_range : list or tuple
        Date range for filtering the dataset [start_date, end_date] (format: ["YYYY-MM-DD", "YYYY-MM-DD"])
    spatial_range : dict
        Dictionary defining the spatial boundaries with keys:
        - 'lat': [min_lat, max_lat]
        - 'lon': [min_lon, max_lon]
     timestep : int or None
        The time index to extract after date selection. 
    convert_longitude : bool, optional (default=True)
        Whether to convert longitude from 0-360 to -180 to 180
    include_wind_speed : bool, optional (default=True)
        Whether to calculate and include wind_speed in the output dataset
    include_attributes : bool, optional (default=True)
        Whether to copy variable attributes from the source dataset
    
    Returns:
    --------
    xarray.Dataset
        A new dataset with processed data
    """
    # Make a copy of the input dataset to avoid modifying the original
    data = ds.copy()
    
    # Extract parameters
    start_date, end_date = date_range
    latitude_range = spatial_range['lat']
    longitude_range = spatial_range['lon']
    
    # Filter dataset for the specified datetime range
    data = data.sel(valid_time=slice(start_date, end_date))
    
    # Convert list of variables to dict if provided as a list
    if isinstance(variables, list):
        var_dict = {var: var for var in variables}
        # Select only the variables we need
        data = data[variables]
    else:
        var_dict = variables
        # Select only the variables we need
        data = data[list(var_dict.keys())]
    
    # Convert longitude from 0-360 to -180 to 180 if requested
    if convert_longitude:
        data = data.assign_coords(longitude=(data.longitude + 180) % 360 - 180)
        # Sort the dataset by the new longitude values
        data = data.sortby('longitude')
    
    # Slice data for the Region of Interest
    data = data.sel(
        latitude=slice(latitude_range[1], latitude_range[0]),
        longitude=slice(longitude_range[0], longitude_range[1])
    ).load()
    
    # If timestep is provided, select that specific timestep
    if timestep is not None:
        processed_data = xr.Dataset(
            coords={
                'longitude': data.longitude,
                'latitude': data.latitude
            }
        )
        
        # Extract each variable at the given timestep
        for src_var, dest_var in var_dict.items():
            if src_var in data:
                var_data = data[src_var].isel(valid_time=timestep)
                processed_data[dest_var] = (['latitude', 'longitude'], var_data.values)
                
                # Copy attributes if requested
                if include_attributes and hasattr(data[src_var], 'attrs'):
                    processed_data[dest_var].attrs = data[src_var].attrs.copy()
                    
        # Calculate and add wind speed if requested and required variables exist
        if include_wind_speed:
            # Check if we need to use the original or renamed variables
            u_var = 'u10' if 'u10' in processed_data else var_dict.get('u10', None)
            v_var = 'v10' if 'v10' in processed_data else var_dict.get('v10', None)
            
            if u_var in processed_data and v_var in processed_data:
                wind_speed = np.sqrt(processed_data[u_var]**2 + processed_data[v_var]**2)
                processed_data['wind_speed'] = (['latitude', 'longitude'], wind_speed.values)
                
                if include_attributes:
                    # Create appropriate attributes for wind_speed
                    processed_data.wind_speed.attrs = {
                        'long_name': 'Wind Speed at 10m',
                        'units': processed_data[u_var].attrs.get('units', 'm s**-1') 
                                if hasattr(processed_data[u_var], 'attrs') and 'units' in processed_data[u_var].attrs 
                                else 'm s**-1'
                    }
    else:
        raise ValueError('Please provide a timestep. ')
    
    # Copy coordinate attributes if requested
    if include_attributes:
        for coord in ['longitude', 'latitude']:
            if coord in data and hasattr(data[coord], 'attrs'):
                processed_data[coord].attrs = data[coord].attrs.copy()
        
        # Copy global attributes
        processed_data.attrs = data.attrs.copy()
    
    return processed_data
```

```python


def subsample_dataset(ds: xr.Dataset, step: int = 2) -> xr.Dataset:
    """
    Subsamples the input xarray Dataset by the specified step along spatial dimensions.
    
    Parameters:
    - ds: xarray.Dataset containing variables including 'lon', 'lat', 'u10', 'v10', and optionally 'wind_speed'.
    - step: int, step size for subsampling.
    
    Returns:
    - subsampled xarray.Dataset
    """
    # Subsample coordinates
    lon_sub = ds['longitude'][::step]
    lat_sub = ds['latitude'][::step]

    # Subsample variables
    u10_sub = ds['u10'].isel(latitude=slice(None, None, step), longitude=slice(None, None, step))
    v10_sub = ds['v10'].isel(latitude=slice(None, None, step), longitude=slice(None, None, step))
    t2m_sub = ds['t2m'].isel(latitude=slice(None, None, step), longitude=slice(None, None, step))
    
    # Handle optional wind_speed variable
    if 'wind_speed' in ds:
        wind_speed_sub = ds['wind_speed'].isel(latitude=slice(None, None, step), longitude=slice(None, None, step))
        return xr.Dataset({
            'u10': u10_sub,
            'v10': v10_sub,
            't2m': t2m_sub,
            'wind_speed': wind_speed_sub,
        
        })
    else:
        return xr.Dataset({
            'u10': u10_sub,
            'v10': v10_sub,
            'lon': lon_sub,
            't2m': t2m_sub,
            'lat': lat_sub
        })
```

```python
def load_boundaries_as_gdf(path: str, verbose:bool=True):
    """
    Load the zip file download from Natural Earth Data as a geodataframe.
    https://www.naturalearthdata.com/downloads/10m-physical-vectors/

    path: The path were the zip file is located
    verbose: Print the head of the dataframe 
    """

    # Create a temporary directory to extract files
    with tempfile.TemporaryDirectory() as temp_dir:
        with ZipFile(path, 'r') as zf:
            # Extract all files into the temporary directory
            zf.extractall(temp_dir)
            
            # Find the .shp file within the extracted contents
            shp_file = [os.path.join(temp_dir, name) for name in zf.namelist() if name.endswith('.shp')][0]
            
            # Load the shapefile into a GeoDataFrame
            gdf = gpd.read_file(shp_file)

    if verbose:
        print(gdf.head())
    return gdf 

# Path to your ZIP file
zip_file_path = "ne_110m_admin_0_countries.zip"

## Create gdf
gdf = load_boundaries_as_gdf(path=zip_file_path,verbose=False)

EEA_list = [
    "Austria", "Belgium", "Bulgaria", "Croatia", "Cyprus", "Czech Republic",
    "Denmark", "Estonia", "Finland", "France", "Germany", "Greece",
    "Hungary", "Ireland", "Italy", "Latvia", "Liechtenstein",
    "Lithuania", "Luxembourg", "Malta", "Netherlands", "Norway", "Poland",
    "Portugal", "Romania", "Slovakia", "Slovenia", "Spain", "Sweden"
]

## Filter countries
filtered_gdf = gdf[gdf["ADMIN"].isin(EEA_list)]


## Handle multipoligons 
filtered_gdf.loc[filtered_gdf['ADMIN']=='France','geometry'] = filtered_gdf[filtered_gdf['ADMIN']=='France'].geometry.iloc[0].geoms[1]
filtered_gdf.loc[filtered_gdf['ADMIN']=='Norway','geometry'] = filtered_gdf.loc[filtered_gdf['ADMIN']=='Norway'].geometry.iloc[0].geoms[1]

## Only keep essencial columns
filtered_gdf = filtered_gdf[['ADMIN','geometry']]
## Parse as geojson
geojson = filtered_gdf[['ADMIN','geometry']].to_json()
```

## Manim Plot

### Manim implementation of Vector Fields for animation pourposes. 
The current implementation below creates a Scene object that contain attributes (gdf and xarray.dataset). The method construct is where the scene is created by default. 
A few bullet points should be concerned:
- Coordinate System: 
    - An Axes object (Mobject Class) is generated and hold the information regarding axes.
    - With the Axes object is possible to place objects within the axes coordinate system instead of using the global coordinate system.
    - The axes object is delimited by two attribute: x_length_ax and y_length_ax.
    - These two attributes will contain the role of bounding box of the canvas regarding the scene. By default, the scene is given by heightXwidth of 8x14.2 (ratio of 16:9).
    - With these two attributes of the class, an axes object is limited by having an arbitrary value of height (y_length_ax) and the width (x_length_ax) is calculated to match the ratio of the xarray.dataset coordinates.
    - This, necessarily implies that the gdf and the xarray should match the same CRS system. <br>
<br>

- Temperature plot:
    - The Temperature plot is a ImageObject from a given imagearray.
    - The ImageObject has its height and width adjusted by the attributes x_length_ax and y_length_ax. <br>
<br> 

- Country Features Outline:
    - Outlines are draw as a Polygon Mobject which their coordinates are being transformed to manim scene coordinates.<br>
<br>

- Vector Field:
    - Vector Field is a Streamline object from the Vector Field mobject

```python
class ERA5VectorField(Scene):
    gdf = None
    data = None

    ## Set up the extension of the axes 
    x_length_ax = None
    y_length_ax = 7
    
    def normalize_array(self, arr):
        """Normalize array to 0–1 range"""
        return (arr - np.min(arr)) / (np.max(arr) - np.min(arr))

    def _geometry_to_mobject(self, geom, coord_func):
        coords = np.array(geom.exterior.coords)
        points = [coord_func(lon, lat) for lon, lat in coords]
        return Polygon(*points, stroke_color=BLACK, stroke_width=1, fill_opacity=0.0)

    def construct(self):
        self.camera.background_color = WHITE

        # Extract temperature data
        lon = self.data.longitude.values
        lat = self.data.latitude.values
        t2m = self.data.t2m.values
        u10 = self.data.u10.values
        v10 = self.data.v10.values

        ## Get the bounding box of the xarray in order to set up Axes
        lon_min, lon_max = np.min(lon), np.max(lon)
        lat_min, lat_max = np.min(lat), np.max(lat)

        # Calculate data aspect ratio
        data_width = lon_max - lon_min
        data_height = lat_max - lat_min
        data_aspect_ratio = data_width / data_height

        # calculate x_length_ax to maintain aspect ratio
        #self.y_length_ax = # Make this larger to fill more screen space
        self.x_length_ax = self.y_length_ax * data_aspect_ratio

        ## ----- AXES -------
        ### It creates a axes coordinate system. 
        #### Every other line plot is done by using axes methods instead of absolute coordinates  
        axes = Axes(
            x_range=[lon_min, lon_max, (lon_max-lon_min)/5], 
            y_range=[lat_min, lat_max, (lat_max-lat_min)/5],
            x_length=self.x_length_ax,
            y_length=self.y_length_ax,
            axis_config={"include_tip": False,
                         "color": DARK_GREY,
                         "font_size":16}
        )
        
        axes_labels = VGroup(
            axes.get_x_axis_label("Lon"),
            axes.get_y_axis_label("Lat")
        )

        # Normalize 
        t2m_norm = self.normalize_array(t2m)
        height, width = t2m_norm.shape

        # Initialize pixel array
        image_array = np.zeros((height, width, 3), dtype=np.uint8)

        for i in range(height):
            for j in range(width):
                val = t2m_norm[i, j]
                rgb = np.array(plasma_r(val)[:3]) * 255 # scale to RGB 0–255
                image_array[i, j] = rgb.astype(np.uint8)

        # image_array = np.flipud(image_array)  # Uncomment in case if Y-axis is inverted

        # Create ImageMobject of temperature 
        image = ImageMobject(image_array)
        
        # Adjust the ImageObject to fit in the axes.
        image.height = self.y_length_ax
        image.width = self.x_length_ax
        
        # Align the image with the axes
        image.move_to(axes.get_center())

        # Title
        title = Text("ERA5 - t2m - Temperature [K]", font_size=24).to_edge(UP)

        # GeoJSON outline coordinate transformer
        def geo_to_manim_coords(lon_val, lat_val):
            # Map from data coordinates to Manim coordinates
            # proportional positioning on the axes
            x_ratio = (lon_val - lon_min) / (lon_max - lon_min)
            y_ratio = (lat_val - lat_min) / (lat_max - lat_min)
            
            # Convert to actual manim coordinates
            x = axes.get_center()[0] - self.x_length_ax/2 + x_ratio * self.x_length_ax
            y = axes.get_center()[1] - self.y_length_ax/2 + y_ratio * self.y_length_ax
            
            return np.array([x, y, 0])

        
        ## Plot Contry Outline - Features 
        europe_outline = VGroup()
        if self.gdf is not None:
            for _, row in self.gdf.iterrows():
                geom = row.geometry
                if geom.geom_type == "Polygon":
                    outline = self._geometry_to_mobject(geom, geo_to_manim_coords)
                    europe_outline.add(outline)
                elif geom.geom_type == "MultiPolygon":
                    for part in geom.geoms:
                        outline = self._geometry_to_mobject(part, geo_to_manim_coords)
                        europe_outline.add(outline)
        

        def vector_field_func(point):
            ## Convert from Manim coordinates to lon/lat indices
            x, y = point[0], point[1]
            
            # Get center 
            axes_center = axes.get_center()
            
            # convert from Manim scene coordinates to relative position in axes
            x_ratio = (x - (axes_center[0] - self.x_length_ax/2)) / self.x_length_ax
            y_ratio = (y - (axes_center[1] - self.y_length_ax/2)) / self.y_length_ax
            
            # convert relative position to data indices
            lon_idx = int(x_ratio * (len(lon) - 1))
            lat_idx = int(y_ratio * (len(lat) - 1))
            
            # ensure indices are within bounds
            lon_idx = max(0, min(lon_idx, len(lon) - 1))
            lat_idx = max(0, min(lat_idx, len(lat) - 1))
            
            # Get wind components
            u = u10[lat_idx, lon_idx]
            v = v10[lat_idx, lon_idx]
            
            # Scale the vector for visualization
            magnitude = np.sqrt(u**2 + v**2)
            scale = 0.5 / (magnitude + 1)  
            
            return np.array([u * scale, v * scale, 0]) 
            
        ## Streamlines
        streamlines = StreamLines(
            vector_field_func,
            x_range=[axes.get_center()[0] - self.x_length_ax/2, axes.get_center()[0] + self.x_length_ax/2, 0.5],
            y_range=[axes.get_center()[1] - self.y_length_ax/2, axes.get_center()[1] + self.y_length_ax/2, 0.5],
            stroke_width=1.5,
            color=BLACK,
        )
        
        # Create a group for everything
        plot_group = Group(image,
                           #axes,
                           #axes_labels,
                           europe_outline)

        # Center the entire group in the frame
        #plot_group.scale_to_fit_height(config.frame_height * 0.9)
        #plot_group.scale_to_fit_width(config.frame_width * 0.75 )
        
        plot_group.center()

        # Add title at the top
        title.next_to(plot_group, UP)
        
        # Add everything to the scene
        self.add(plot_group, title, streamlines)
        
        #Animation 
        #self.add(streamlines)
        streamlines.start_animation(warm_up=False,
                                    flow_speed=1.5,
                                    time_width=0.7)
        self.wait(4)
        self.play(streamlines.end_animation()) ##fadeout 


# Attach mock data
ERA5VectorField.gdf = filtered_gdf
ERA5VectorField.data = sub2_dataset
```

### Modified Class

The main difference of this class to the above one is regarding the outline features. In the modified class the outline countries features are implemented as a method of the Axes object. Constrasting with the class above were the countries were implemented with the Polygon object and had to be transformed from global coordinates to the scene coordinates. On this approach, the features are handled within the axes boundaries and Axes coordinates. 

```python
class ERA5VectorField_mod(Scene):
    gdf = None
    data = None

    ## Set up the extension of the axes 
    x_length_ax = None
    y_length_ax = 7
    
    def normalize_array(self, arr):
        """Normalize array to 0–1 range"""
        return (arr - np.min(arr)) / (np.max(arr) - np.min(arr))

    def construct(self):
        self.camera.background_color = WHITE

        # Extract temperature data
        lon = self.data.longitude.values
        lat = self.data.latitude.values
        t2m = self.data.t2m.values
        u10 = self.data.u10.values
        v10 = self.data.v10.values

        ## Get the bounding box of the xarray in order to set up Axes
        lon_min, lon_max = np.min(lon), np.max(lon)
        lat_min, lat_max = np.min(lat), np.max(lat)

        # Calculate data aspect ratio
        data_width = lon_max - lon_min
        data_height = lat_max - lat_min
        data_aspect_ratio = data_width / data_height

        # calculate x_length_ax to maintain aspect ratio
        #self.y_length_ax = # Make this larger to fill more screen space
        self.x_length_ax = self.y_length_ax * data_aspect_ratio

        ## ----- AXES -------
        ### It creates a axes coordinate system. 
        #### Every other line plot is done by using axes methods instead of absolute coordinates  
        axes = Axes(
            x_range=[lon_min, lon_max, (lon_max-lon_min)/5], 
            y_range=[lat_min, lat_max, (lat_max-lat_min)/5],
            x_length=self.x_length_ax,
            y_length=self.y_length_ax,
            axis_config={"include_tip": False,
                         "color": DARK_GREY,
                         "font_size":16}
        )
        
        axes_labels = VGroup(
            axes.get_x_axis_label("Lon"),
            axes.get_y_axis_label("Lat")
        )

        # Normalize 
        t2m_norm = self.normalize_array(t2m)
        height, width = t2m_norm.shape

        # Initialize pixel array
        image_array = np.zeros((height, width, 3), dtype=np.uint8)

        for i in range(height):
            for j in range(width):
                val = t2m_norm[i, j]
                rgb = np.array(plasma_r(val)[:3]) * 255 # scale to RGB 0–255
                image_array[i, j] = rgb.astype(np.uint8)

        # image_array = np.flipud(image_array)  # Uncomment in case if Y-axis is inverted

        # Create ImageMobject of temperature 
        image = ImageMobject(image_array)
        
        # Adjust the ImageObject to fit in the axes.
        image.height = self.y_length_ax
        image.width = self.x_length_ax
        
        # Align the image with the axes
        image.move_to(axes.get_center())

        # Title
        title = Text("ERA5 - t2m - Temperature [K]", font_size=24).to_edge(UP)

        def plot_geojson_on_axes(axes, gdf):
            geo_lines = VGroup()
            for _, row in gdf.iterrows():
                geom = row.geometry
                if geom.geom_type == "Polygon":
                    coords = np.array(geom.exterior.coords)
                    x, y = coords[:, 0], coords[:, 1]
                    line = axes.plot_line_graph(x, y, add_vertex_dots=False, line_color=BLACK, stroke_width=1)
                    geo_lines.add(line)
                elif geom.geom_type == "MultiPolygon":
                    for part in geom.geoms:
                        coords = np.array(part.exterior.coords)
                        x, y = coords[:, 0], coords[:, 1]
                        line = axes.plot_line_graph(x, y, add_vertex_dots=False, line_color=BLACK, stroke_width=1)
                        geo_lines.add(line)
            return geo_lines
            
        ## Plot Contry Outline - Features 
        europe_outline = VGroup()
        if self.gdf is not None:
            europe_outline = plot_geojson_on_axes(axes, self.gdf)
        

        def vector_field_func(point):
            ## Convert from Manim coordinates to lon/lat indices
            x, y = point[0], point[1]
            
            # Get center 
            axes_center = axes.get_center()
            
            # convert from Manim scene coordinates to relative position in axes
            x_ratio = (x - (axes_center[0] - self.x_length_ax/2)) / self.x_length_ax
            y_ratio = (y - (axes_center[1] - self.y_length_ax/2)) / self.y_length_ax
            
            # convert relative position to data indices
            lon_idx = int(x_ratio * (len(lon) - 1))
            lat_idx = int(y_ratio * (len(lat) - 1))
            
            # ensure indices are within bounds
            lon_idx = max(0, min(lon_idx, len(lon) - 1))
            lat_idx = max(0, min(lat_idx, len(lat) - 1))
            
            # Get wind components
            u = u10[lat_idx, lon_idx]
            v = v10[lat_idx, lon_idx]
            
            # Scale the vector for visualization
            magnitude = np.sqrt(u**2 + v**2)
            scale = 0.5 / (magnitude + 1)  
            
            return np.array([u * scale, v * scale, 0]) 
            
        ## Streamlines
        streamlines = StreamLines(
            vector_field_func,
            x_range=[axes.get_center()[0] - self.x_length_ax/2, axes.get_center()[0] + self.x_length_ax/2, 0.5],
            y_range=[axes.get_center()[1] - self.y_length_ax/2, axes.get_center()[1] + self.y_length_ax/2, 0.5],
            stroke_width=1.5,
            color=BLACK,
            padding=1,
            opacity = 0.85,
            max_anchors_per_line=75
        )
        
        # Create a group for everything
        plot_group = Group(image,
                           #axes,
                           #axes_labels,
                           europe_outline)

        # Center the entire group in the frame
        #plot_group.scale_to_fit_height(config.frame_height * 0.9)
        #plot_group.scale_to_fit_width(config.frame_width * 0.75 )
        
        plot_group.center()

        # Add title at the top
        title.next_to(plot_group, UP)
        
        # Add everything to the scene
        self.add(plot_group, title, streamlines)
        
        #Animation 
        #self.add(streamlines)
        streamlines.start_animation(warm_up=False,
                                    flow_speed=1.5,
                                    time_width=0.7)
        self.wait(4)
        
        #self.play(streamlines.end_animation()) ##fadeout 


# Attach mock data
ERA5VectorField_mod.gdf = filtered_gdf
ERA5VectorField_mod.data = sub2_dataset
```

```python

class ERA5AnimationRenderer(Scene):
    def construct(self):
        # This is a simpler version that focuses just on the animation
        # Create mock data for demonstration
        nx, ny = 50, 50
        x = np.linspace(-5, 5, nx)
        y = np.linspace(-3, 3, ny)
        X, Y = np.meshgrid(x, y)
        
        # Create a circular flow pattern
        cx, cy = 0, 0  # Center of circulation
        r = np.sqrt((X - cx)**2 + (Y - cy)**2)
        theta = np.arctan2(Y - cy, X - cx)
        
        # Compute vector components
        u = -np.sin(theta) * (1 / (r + 0.5))
        v = np.cos(theta) * (1 / (r + 0.5))
        
        # Add some secondary patterns
        u += 0.3 * np.sin(X * 2) * np.cos(Y * 2)
        v += 0.3 * np.cos(X * 2) * np.sin(Y * 2)
        
        # Create a temperature-like gradient
        temp = 20 - Y * 2 + np.sin(X * 1.5) * 2
        
        # Normalize temperature for color mapping
        temp_norm = (temp - temp.min()) / (temp.max() - temp.min())
        
        # Create the background heatmap
        heatmap_colors = []
        for i in range(ny):
            row = []
            for j in range(nx):
                # Create a plasma-like colormap: BLUE -> PURPLE -> RED -> YELLOW
                val = temp_norm[i, j]
                if val < 0.25:
                    color = interpolate_color(BLUE, PURPLE, val*4)
                elif val < 0.5:
                    color = interpolate_color(PURPLE, RED, (val-0.25)*4)
                elif val < 0.75:
                    color = interpolate_color(RED, ORANGE, (val-0.5)*4)
                else:
                    color = interpolate_color(ORANGE, YELLOW, (val-0.75)*4)
                row.append(color)
            heatmap_colors.append(row)
        
        # Create a pixel array for heatmap
        pixels = np.zeros((ny, nx, 3))
        for i in range(ny):
            for j in range(nx):
                pixels[i, j] = color_to_rgb(heatmap_colors[i][j])
        
        # Create and position the heatmap
        heatmap = ImageMobject(pixels).scale(3)
        
        # Define the vector field function
        def vector_field_func(point):
            # Get x, y coordinates
            px, py = point[0], point[1]
            
            # Find nearest indices in our grid
            ix = int((px + 5) * (nx-1) / 10)
            iy = int((py + 3) * (ny-1) / 6)
            
            # Clamp to valid indices
            ix = max(0, min(ix, nx-1))
            iy = max(0, min(iy, ny-1))
            
            # Get vector components
            ux = u[iy, ix]
            vy = v[iy, ix]
            
            # Scale for visibility
            scale = 0.3
            return np.array([ux * scale, vy * scale, 0])
        
        # Create streamlines
        streamlines = StreamLines(
            vector_field_func,
            x_range=[-5, 5, 0.5],
            y_range=[-3, 3, 0.5],
            stroke_width=2,
            color=WHITE,
        )
        
        # Create axes and borders
        axes = Axes(
            x_range=[-5, 5, 1],
            y_range=[-3, 3, 1],
            x_length=10,
            y_length=6,
            axis_config={"color": DARK_GREY, "include_ticks": True}
        )
        
        # Add labels
        x_label = Text("Longitude", font_size=24).next_to(axes, DOWN)
        y_label = Text("Latitude", font_size=24).rotate(PI/2).next_to(axes, LEFT)

        # Custom vertical colorbar using stacked rectangles
        num_colors = 50
        colorbar_colors = color_gradient([BLUE, PURPLE, RED, ORANGE, YELLOW], num_colors)
        colorbar_height = 6
        colorbar_width = 0.3
        rectangles = VGroup()

        for i in range(num_colors):
            rect = Rectangle(
                width=colorbar_width,
                height=colorbar_height / num_colors,
                fill_color=colorbar_colors[i],
                fill_opacity=1,
                stroke_width=0
            )
            rect.move_to(UP * (i - num_colors/2 + 0.5) * (colorbar_height / num_colors))
            rectangles.add(rect)

        colorbar = rectangles
        colorbar.move_to([5.5, 0, 0])  # Adjust position if needed
        colorbar_label = Text("[K]", font_size=24).next_to(colorbar, RIGHT, buff=0.2)

        
        # Add title
        title = Text("ERA5 - t2m - Temperature [K]", font_size=36).to_edge(UP)

        ## Build the scene 
        self.add(heatmap)
        self.add(axes, x_label, y_label)
        self.add(colorbar, colorbar_label)
        self.add(title)
        self.add(streamlines)

        # Animate streamlines
        streamlines.start_animation(flow_speed=1.5, time_width=0.5)

        # Let the animation play for 10 seconds
        self.wait(10)


# To run this animation, uncomment one of these lines:
# manim -pql ERA5VectorFieldAnimation.py ERA5VectorFieldAnimation
#%manim -pql ERA5AnimationRenderer.py ERA5AnimationRenderer
```

```python
class ERA5VectorFieldAnimation(Scene):
    ## Create class attributes
    gdf = None
    data = None 

    ##Methods
    def plasma_r_colormap(self, val):
        return plasma_r(val)
    
    def color_to_rgb(self, color):
        return mcolors.to_rgb(color)

    ## Construct 
    def construct(self):
        self.camera.background_color = WHITE

        ## Extract values
        lon = self.data.longitude.values
        lat = self.data.latitude.values
        u10 = self.data.u10.values
        v10 = self.data.v10.values
        t2m = self.data.t2m.values

        # Axes setup
        axes = Axes(
            x_range=[-10, 30, 5],
            y_range=[35, 70, 5],
            x_length=10,
            y_length=5,
            axis_config={"include_tip": False, "color": DARK_GREY}
        )
        axes_labels = VGroup(
            axes.get_x_axis_label("Longitude"),
            axes.get_y_axis_label("Latitude")
        )
        title = Text("ERA5 - t2m - Temperature [K]", font_size=36).to_edge(UP)

        ### ------ T2m Heatmap ----------
        # normalize
        t2m_norm = (t2m - t2m.min()) / (t2m.max() - t2m.min())
        height, width = t2m_norm.shape
        pixel_array = np.zeros((height, width, 3))
        
        # iterate through matrix
        for i in range(height):
            for j in range(width):
                val = t2m_norm[i, j]
                pixel_array[i, j] = self.color_to_rgb(self.plasma_r_colormap(val))
        
        # Create imageMobject
        heatmap = ImageMobject(pixel_array)
        heatmap.height = 6
        heatmap.width = 10
        heatmap.move_to(axes.get_center())

        # ------ Colorbar setup ----
        colorbar = Rectangle(height=6, width=0.3, fill_opacity=1).next_to(axes, RIGHT, buff=0.5)
        colorbar.set_color_by_gradient(DARK_BLUE, PURPLE, YELLOW)
        colorbar_label = Text("[K]", font_size=24).next_to(colorbar, RIGHT)

        # --- Coordinate mapper ---
        def geo_to_manim_coords(lon_val, lat_val):
            x = (lon_val - lon.min()) / (lon.max() - lon.min()) * 10 - 5
            y = (lat_val - lat.min()) / (lat.max() - lat.min()) * 6 - 3
            return np.array([x, y, 0])

        # --- Add GeoJSON outline ---
        europe_outline = VGroup()
        for _, row in self.gdf.iterrows():
            geom = row.geometry
            if geom.geom_type == "Polygon":
                outline = self._geometry_to_mobject(geom, geo_to_manim_coords)
                europe_outline.add(outline)
            elif geom.geom_type == "MultiPolygon":
                for part in geom.geoms:
                    outline = self._geometry_to_mobject(part, geo_to_manim_coords)
                    europe_outline.add(outline)

        # --- Vector field function ---
        def vector_field_func(point):
            x, y = point[0], point[1]
            lon_idx = int((x + 5) / 10 * (len(lon) - 1))
            lat_idx = int((y + 3) / 6 * (len(lat) - 1))
            lon_idx = max(0, min(lon_idx, len(lon) - 1))
            lat_idx = max(0, min(lat_idx, len(lat) - 1))
            u = u10[lat_idx, lon_idx]
            v = v10[lat_idx, lon_idx]
            magnitude = np.sqrt(u**2 + v**2)
            scale = 0.3 / (magnitude + 1)
            return np.array([u * scale, v * scale, 0])

        # Create the vector field
        # According to the documentation, VectorField doesn't take x_range or y_range
        vector_field = VectorField(
            vector_field_func
        )
        
        # For StreamLines - x_range and y_range should be [min, max, delta]
        stream_lines = StreamLines(
            vector_field_func,
            x_range=[-5, 5, 0.3],  # min, max, delta_x
            y_range=[-3, 3, 0.3],  # min, max, delta_y
            stroke_width=1.5,
            dt=0.05,
            virtual_time=3,
            max_anchors_per_line=100
        )

        # --- Add to scene ---
        self.add(heatmap, axes, axes_labels, title, europe_outline)
        self.add(colorbar, colorbar_label)
        self.add(stream_lines)
        stream_lines.start_animation(flow_speed=1.5, time_width=0.5)
        self.wait(10)
        self.add(vector_field)
        self.wait(5)
        

    def _geometry_to_mobject(self, geom, coord_func):
        coords = np.array(geom.exterior.coords)
        points = [coord_func(lon, lat) for lon, lat in coords]
        return Polygon(*points, stroke_color=GREY, stroke_width=1.5, fill_opacity=0.0)

## Initialize the class
ERA5VectorFieldAnimation.gdf = filtered_gdf
ERA5VectorFieldAnimation.data = create_mock_era5_data()
```

```python
# Import necessary libraries
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.cm import plasma_r
import matplotlib.colors as mcolors
from manim import *

class ERA5VectorFieldAnimation(Scene):
    ## Create class attributes
    gdf = None
    data = None 
    
    ## Set up variables 
    x_length_ax = 10
    y_length_ax = 5
    
    ## ------Methods  ----
    def plasma_r_colormap(self, val):
        """ Returns a RGBA tuple for a given value
        """
        return plasma_r(val)
    
    def color_to_rgb(self, color):
        """ Converts a given RGBA tuple into a RGB
        """
        return mcolors.to_rgb(color)

    def normalize_array(self, arr):
        """ Normalize a given array 
        """
        norm_arr = (arr - np.min(arr)) / (np.max(arr) - np.min(arr))
        return norm_arr
        
    ## ------- Construct --------
    def construct(self):
        self.camera.background_color = WHITE

        ## Extract values
        lon = self.data.longitude.values
        lat = self.data.latitude.values
        u10 = self.data.u10.values
        v10 = self.data.v10.values
        t2m = self.data.t2m.values

        # Define temperature range for consistent coloring
        t2m_min = np.min(t2m)
        t2m_max = np.max(t2m)

        # Axes setup
        axes = Axes(
            x_range=[-10, 30, 5], ## limits of the scene
            y_range=[35, 70, 5], ## limits of the scene representing Europe 
            x_length=self.x_length_ax, 
            y_length=self.y_length_ax,
            axis_config={"include_tip": False, "color": DARK_GREY}
        )
        axes_labels = VGroup(
            axes.get_x_axis_label("Longitude"),
            axes.get_y_axis_label("Latitude")
        )
        self.add(axes, axes_labels)
        
        title = Text("ERA5 - t2m - Temperature [K]", font_size=36).to_edge(UP)

        ### ------ Temperature Heatmap ----------
        # normalize
        t2m_norm = self.normalize_array(t2m)
        
        height, width = t2m_norm.shape
        
       # Create the pixel array using np.uint8 for 8-bit RGB image
        pixel_array = np.zeros((height, width, 3), dtype=np.uint8)
        cmap = plasma_r
        
        # Populate the array with colormap values scaled to 0–255
        for i in range(height):
            for j in range(width):
                val = t2m_norm[i, j]
                rgb = np.array(cmap(val)[:3]) * 255  # Convert to 0–255 range
                pixel_array[i, j] = rgb.astype(np.uint8)
        
        # Create imageMobject
        heatmap = ImageMobject(pixel_array)
        heatmap.height = self.y_length_ax
        heatmap.width = self.x_length_ax
        
        # Position the heatmap properly on the axes
        heatmap.move_to(axes.get_center())

        # ------ Colorbar setup ----
        ## Use the same colormap as the heatmap
        num_colors = 100  # More colors for smoother gradient
        colorbar_height = 5
        colorbar_width = 0.3
        
        # Create colors using the same colormap
        colorbar_values = np.linspace(0, 1, num_colors)
        rectangles = VGroup()

        for i, val in enumerate(colorbar_values):
            color_rgb = cmap(val)[:3]  # Get RGB values
            rect = Rectangle(
                width=colorbar_width,
                height=colorbar_height / num_colors,
                fill_color=rgb_to_color(color_rgb),  # Convert RGB to Manim color
                fill_opacity=1,
                stroke_width=0
            )
            rect.move_to(UP * (i - num_colors/2 + 0.5) * (colorbar_height / num_colors))
            rectangles.add(rect)

        colorbar = rectangles
        colorbar.move_to([6, 0, 0])  # Adjusted position
        
        # Add min and max temperature labels
        min_temp = Text(f"{t2m_min:.1f} K", font_size=20).next_to(colorbar, DOWN, buff=0.2)
        max_temp = Text(f"{t2m_max:.1f} K", font_size=20).next_to(colorbar, UP, buff=0.2)
        colorbar_title = Text("Temperature", font_size=24).next_to(colorbar, RIGHT, buff=0.3)

        # --- Coordinate mapper ---
        def geo_to_manim_coords(lon_val, lat_val):
            # Map longitude and latitude to axes coordinates
            x = axes.c2p(lon_val, 0)[0]
            y = axes.c2p(0, lat_val)[1]
            return np.array([x, y, 0])

        # --- Add GeoJSON outline with better positioning ---
        europe_outline = VGroup()
        if self.gdf is not None:
            for _, row in self.gdf.iterrows():
                geom = row.geometry
                if geom.geom_type == "Polygon":
                    outline = self._geometry_to_mobject(geom, geo_to_manim_coords)
                    europe_outline.add(outline)
                elif geom.geom_type == "MultiPolygon":
                    for part in geom.geoms:
                        outline = self._geometry_to_mobject(part, geo_to_manim_coords)
                        europe_outline.add(outline)
                    
        # --- Add to scene ---
        self.add(heatmap)  # Add heatmap first (below outlines)
        self.add(europe_outline)  # Add outlines on top
        self.add(colorbar, colorbar_title, min_temp, max_temp)
        self.add(title)
        self.wait(10)
        
    def _geometry_to_mobject(self, geom, coord_func):
        coords = np.array(geom.exterior.coords)
        points = [coord_func(lon, lat) for lon, lat in coords]
        return Polygon(*points, stroke_color=BLACK, stroke_width=1, fill_opacity=0.0)

def rgb_to_color(rgb_array):
    """Convert RGB array (values 0-1) to Manim color"""
    # Convert RGB values to hex color string
    r, g, b = rgb_array
    return rgb_to_hex([r, g, b])

## Initialize the class
ERA5VectorFieldAnimation.gdf = filtered_gdf
ERA5VectorFieldAnimation.data = create_mock_era5_data()
```

```python
class ERA5TempAsImage(Scene):
    gdf = None
    data = None

    ## Image size of the Manim Scene Units
    # Keeping a 16:9 aspect ratio
    # The default Scene screen resolution is 1920x1080 given by config["pixel_height"]
    x_length_ax = 14.2
    y_length_ax = 8  #y-axis spans 10 vertical units

    def normalize_array(self, arr):
        """Normalize array to 0–1 range"""
        return (arr - np.min(arr)) / (np.max(arr) - np.min(arr))

    def construct(self):
        self.camera.background_color = WHITE

        # Extract temperature data
        lon = self.data.longitude.values
        lat = self.data.latitude.values
        t2m = self.data.t2m.values

        ## Get the bounding box of the xarray in order to set up Axes
        lon_min, lon_max = np.min(lon), np.max(lon)
        lat_min, lat_max = np.min(lat), np.max(lat)

        # Calculate data aspect ratio
        data_width = lon_max - lon_min
        data_height = lat_max - lat_min
        data_aspect_ratio = data_width / data_height

        # Set y_length_ax and calculate x_length_ax to maintain aspect ratio
        self.y_length_ax = 5  # You can adjust this base size
        self.x_length_ax = self.y_length_ax * data_aspect_ratio

        ## ---AXES
        axes = Axes(
            x_range=[lon_min, lon_max, 5], #guarantee the bbox matches xarray bbox
            y_range=[lat_min, lat_max, 5],
            x_length=self.x_length_ax,
            y_length=self.y_length_ax,
            axis_config={"include_tip": False, "color": DARK_GREY}
        )
        axes_labels = VGroup(
            axes.get_x_axis_label("Longitude"),
            axes.get_y_axis_label("Latitude")
        )

        # Normalize to 0–1 then scale to RGB 0–255
        t2m_norm = self.normalize_array(t2m)
        height, width = t2m_norm.shape

        # Initialize pixel array
        image_array = np.zeros((height, width, 3), dtype=np.uint8)

        for i in range(height):
            for j in range(width):
                val = t2m_norm[i, j]
                rgb = np.array(plasma_r(val)[:3]) * 255
                image_array[i, j] = rgb.astype(np.uint8)

        # Convert to ImageMobject
        image = ImageMobject(image_array)
        image.height = self.y_length_ax
        image.width = self.x_length_ax
        image.move_to(axes.get_center())

        # Title
        title = Text("ERA5 - t2m - Temperature [K]", font_size=36).to_edge(UP)

        # GeoJSON outline (optional)
        def geo_to_manim_coords(lon_val, lat_val):
            x = axes.c2p(lon_val, 0)[0]
            y = axes.c2p(0, lat_val)[1]
            return np.array([x, y, 0])

        europe_outline = VGroup()
        if self.gdf is not None:
            for _, row in self.gdf.iterrows():
                geom = row.geometry
                if geom.geom_type == "Polygon":
                    outline = self._geometry_to_mobject(geom, geo_to_manim_coords)
                    europe_outline.add(outline)
                elif geom.geom_type == "MultiPolygon":
                    for part in geom.geoms:
                        outline = self._geometry_to_mobject(part, geo_to_manim_coords)
                        europe_outline.add(outline)

        #cartesian plot 
        numberplane = NumberPlane()
        # Assemble
        image.move_to(axes.get_center())
        self.add(image, axes, axes_labels, europe_outline, title)
        self.add(numberplane)
        #self.wait(10)

    def _geometry_to_mobject(self, geom, coord_func):
        coords = np.array(geom.exterior.coords)
        points = [coord_func(lon, lat) for lon, lat in coords]
        return Polygon(*points, stroke_color=BLACK, stroke_width=1, fill_opacity=0.0)

# Attach mock data
ERA5TempAsImage.gdf = filtered_gdf
ERA5TempAsImage.data = create_mock_era5_data()
```

```python
class ERA5TempAsImage(Scene):
    gdf = None
    data = None
    
    ## Image size of the Manim Scene Units
    # Keeping a 16:9 aspect ratio
    # The default Scene screen resolution is 1920x1080 given by config["pixel_height"]

    ## Set up the extension of the axes 
    x_length_ax = None
    y_length_ax = 7.5
    
    def normalize_array(self, arr):
        """Normalize array to 0–1 range"""
        return (arr - np.min(arr)) / (np.max(arr) - np.min(arr))

    def _geometry_to_mobject(self, geom, coord_func):
        coords = np.array(geom.exterior.coords)
        points = [coord_func(lon, lat) for lon, lat in coords]
        return Polygon(*points, stroke_color=BLACK, stroke_width=1, fill_opacity=0.0)

    def construct(self):
        self.camera.background_color = WHITE

        # Extract temperature data
        lon = self.data.longitude.values
        lat = self.data.latitude.values
        t2m = self.data.t2m.values

        ## Get the bounding box of the xarray in order to set up Axes
        lon_min, lon_max = np.min(lon), np.max(lon)
        lat_min, lat_max = np.min(lat), np.max(lat)

        # Calculate data aspect ratio
        data_width = lon_max - lon_min
        data_height = lat_max - lat_min
        data_aspect_ratio = data_width / data_height

        # calculate x_length_ax to maintain aspect ratio
        #self.y_length_ax = # Make this larger to fill more screen space
        self.x_length_ax = self.y_length_ax * data_aspect_ratio

        ## ----- AXES -------
        ### It creates a axes coordinate system. 
        #### Every other line plot is done by using axes methods instead of absolute coordinates  
        axes = Axes(
            x_range=[lon_min, lon_max, (lon_max-lon_min)/5], 
            y_range=[lat_min, lat_max, (lat_max-lat_min)/5],
            x_length=self.x_length_ax,
            y_length=self.y_length_ax,
            axis_config={"include_tip": False,
                         "color": DARK_GREY,
                         "font_size":16}
        )
        
        axes_labels = VGroup(
            axes.get_x_axis_label("Lon"),
            axes.get_y_axis_label("Lat")
        )

        # Normalize 
        t2m_norm = self.normalize_array(t2m)
        height, width = t2m_norm.shape

        # Initialize pixel array
        image_array = np.zeros((height, width, 3), dtype=np.uint8)

        for i in range(height):
            for j in range(width):
                val = t2m_norm[i, j]
                rgb = np.array(plasma_r(val)[:3]) * 255 # scale to RGB 0–255
                image_array[i, j] = rgb.astype(np.uint8)

        # image_array = np.flipud(image_array)  # Uncomment in case if Y-axis is inverted

        # Create ImageMobject of temperature 
        image = ImageMobject(image_array)
        
        # Adjust the ImageObject to fit in the axes.
        image.height = self.y_length_ax
        image.width = self.x_length_ax
        
        # Align the image with the axes
        image.move_to(axes.get_center())

        # Title
        title = Text("ERA5 - t2m - Temperature [K]", font_size=24).to_edge(UP)

        # GeoJSON outline coordinate transformer
        def geo_to_manim_coords(lon_val, lat_val):
            # Map from data coordinates to Manim coordinates
            # proportional positioning on the axes
            x_ratio = (lon_val - lon_min) / (lon_max - lon_min)
            y_ratio = (lat_val - lat_min) / (lat_max - lat_min)
            
            # Convert to actual manim coordinates
            x = axes.get_center()[0] - self.x_length_ax/2 + x_ratio * self.x_length_ax
            y = axes.get_center()[1] - self.y_length_ax/2 + y_ratio * self.y_length_ax
            
            return np.array([x, y, 0])

        

        europe_outline = VGroup()
        if self.gdf is not None:
            for _, row in self.gdf.iterrows():
                geom = row.geometry
                if geom.geom_type == "Polygon":
                    outline = self._geometry_to_mobject(geom, geo_to_manim_coords)
                    europe_outline.add(outline)
                elif geom.geom_type == "MultiPolygon":
                    for part in geom.geoms:
                        outline = self._geometry_to_mobject(part, geo_to_manim_coords)
                        europe_outline.add(outline)
        
        # Create a group for everything
        plot_group = Group(image,
                           axes,
                           #axes_labels,
                           europe_outline)
        
        # Center the entire group in the frame
        #plot_group.scale_to_fit_height(config.frame_height * 0.9)
        #plot_group.scale_to_fit_width(config.frame_width * 0.75 )
        plot_group.center()
        
        # Add title at the top
        title.next_to(plot_group, UP)
        
        # Add everything to the scene
        self.add(plot_group, title)
        #self.wait(5)




# Attach mock data
ERA5TempAsImage.gdf = filtered_gdf
ERA5TempAsImage.data = create_mock_era5_data()
```

```python
class ERA5TempAsImage(Scene):
    gdf = None
    data = None

    ## Set up the extension of the axes 
    x_length_ax = None
    y_length_ax = 7
    
    def normalize_array(self, arr):
        """Normalize array to 0–1 range"""
        return (arr - np.min(arr)) / (np.max(arr) - np.min(arr))

    def _geometry_to_mobject(self, geom, coord_func):
        coords = np.array(geom.exterior.coords)
        points = [coord_func(lon, lat) for lon, lat in coords]
        return Polygon(*points, stroke_color=BLACK, stroke_width=1, fill_opacity=0.0)

    def construct(self):
        self.camera.background_color = WHITE

        # Extract temperature data
        lon = self.data.longitude.values
        lat = self.data.latitude.values
        t2m = self.data.t2m.values
        u10 = self.data.u10.values
        v10 = self.data.v10.values

        ## Get the bounding box of the xarray in order to set up Axes
        lon_min, lon_max = np.min(lon), np.max(lon)
        lat_min, lat_max = np.min(lat), np.max(lat)

        # Calculate data aspect ratio
        data_width = lon_max - lon_min
        data_height = lat_max - lat_min
        data_aspect_ratio = data_width / data_height

        # calculate x_length_ax to maintain aspect ratio
        #self.y_length_ax = # Make this larger to fill more screen space
        self.x_length_ax = self.y_length_ax * data_aspect_ratio

        ## ----- AXES -------
        ### It creates a axes coordinate system. 
        #### Every other line plot is done by using axes methods instead of absolute coordinates  
        axes = Axes(
            x_range=[lon_min, lon_max, (lon_max-lon_min)/5], 
            y_range=[lat_min, lat_max, (lat_max-lat_min)/5],
            x_length=self.x_length_ax,
            y_length=self.y_length_ax,
            axis_config={"include_tip": False,
                         "color": DARK_GREY,
                         "font_size":16}
        )
        
        axes_labels = VGroup(
            axes.get_x_axis_label("Lon"),
            axes.get_y_axis_label("Lat")
        )

        # Normalize 
        t2m_norm = self.normalize_array(t2m)
        height, width = t2m_norm.shape

        # Initialize pixel array
        image_array = np.zeros((height, width, 3), dtype=np.uint8)

        for i in range(height):
            for j in range(width):
                val = t2m_norm[i, j]
                rgb = np.array(plasma_r(val)[:3]) * 255 # scale to RGB 0–255
                image_array[i, j] = rgb.astype(np.uint8)

        # image_array = np.flipud(image_array)  # Uncomment in case if Y-axis is inverted

        # Create ImageMobject of temperature 
        image = ImageMobject(image_array)
        
        # Adjust the ImageObject to fit in the axes.
        image.height = self.y_length_ax
        image.width = self.x_length_ax
        
        # Align the image with the axes
        image.move_to(axes.get_center())

        # Title
        title = Text("ERA5 - t2m - Temperature [K]", font_size=24).to_edge(UP)

        # GeoJSON outline coordinate transformer
        def geo_to_manim_coords(lon_val, lat_val):
            # Map from data coordinates to Manim coordinates
            # proportional positioning on the axes
            x_ratio = (lon_val - lon_min) / (lon_max - lon_min)
            y_ratio = (lat_val - lat_min) / (lat_max - lat_min)
            
            # Convert to actual manim coordinates
            x = axes.get_center()[0] - self.x_length_ax/2 + x_ratio * self.x_length_ax
            y = axes.get_center()[1] - self.y_length_ax/2 + y_ratio * self.y_length_ax
            
            return np.array([x, y, 0])

        
        ## Plot Contry Outline - Features 
        europe_outline = VGroup()
        if self.gdf is not None:
            for _, row in self.gdf.iterrows():
                geom = row.geometry
                if geom.geom_type == "Polygon":
                    outline = self._geometry_to_mobject(geom, geo_to_manim_coords)
                    europe_outline.add(outline)
                elif geom.geom_type == "MultiPolygon":
                    for part in geom.geoms:
                        outline = self._geometry_to_mobject(part, geo_to_manim_coords)
                        europe_outline.add(outline)
        

        def vector_field_func(point):
            ## Convert from Manim coordinates to lon/lat indices
            x, y = point[0], point[1]
            
            # Get center 
            axes_center = axes.get_center()
            
            # convert from Manim scene coordinates to relative position in axes
            x_ratio = (x - (axes_center[0] - self.x_length_ax/2)) / self.x_length_ax
            y_ratio = (y - (axes_center[1] - self.y_length_ax/2)) / self.y_length_ax
            
            # convert relative position to data indices
            lon_idx = int(x_ratio * (len(lon) - 1))
            lat_idx = int(y_ratio * (len(lat) - 1))
            
            # ensure indices are within bounds
            lon_idx = max(0, min(lon_idx, len(lon) - 1))
            lat_idx = max(0, min(lat_idx, len(lat) - 1))
            
            # Get wind components
            u = u10[lat_idx, lon_idx]
            v = v10[lat_idx, lon_idx]
            
            # Scale the vector for visualization
            magnitude = np.sqrt(u**2 + v**2)
            scale = 0.5 / (magnitude + 1)  
            
            return np.array([u * scale, v * scale, 0]) 
            
        ## Streamlines
        streamlines = StreamLines(
            vector_field_func,
            x_range=[axes.get_center()[0] - self.x_length_ax/2, axes.get_center()[0] + self.x_length_ax/2, 0.5],
            y_range=[axes.get_center()[1] - self.y_length_ax/2, axes.get_center()[1] + self.y_length_ax/2, 0.5],
            stroke_width=1.5,
            color=BLACK,
        )
        
        # Create a group for everything
        plot_group = Group(image,
                           axes,
                           #axes_labels,
                           europe_outline)

        # Center the entire group in the frame
        #plot_group.scale_to_fit_height(config.frame_height * 0.9)
        #plot_group.scale_to_fit_width(config.frame_width * 0.75 )
        
        plot_group.center()

        # Add title at the top
        title.next_to(plot_group, UP)
        
        # Add everything to the scene
        self.add(plot_group, title, streamlines)
        
        #Animation 
        #self.add(streamlines)
        streamlines.start_animation(warm_up=False,
                                    flow_speed=1.5,
                                    time_width=0.7)
        self.wait(4)
        self.play(streamlines.end_animation()) ##fadeout 


# Attach mock data
ERA5TempAsImage.gdf = filtered_gdf
ERA5TempAsImage.data = create_mock_era5_data()
```
